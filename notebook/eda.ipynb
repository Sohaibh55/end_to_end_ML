#### Convert Grades to numeric data

```python
grade_columns = [
    "math_G1",
    "math_G2",
    "por_G1",
    "por_G2"
]

for column in grade_columns:
    df[column] = df[column].str.replace('"', '', regex=False)
    df[column] = pd.to_numeric(df[column], errors="coerce")
    
```

```python
print(df[grade_columns].head())
print()
print(df[grade_columns].dtypes)
```

##### Label encoding 
  - convert Sex {F,M} to 0 for men and 1 for women 

```python
df['sex'] = df['sex'].str.replace('""' , '' ,regex=False)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
print(df["sex"].value_counts(dropna=False))
print(df["sex"].unique())

print(df["school"].value_counts(dropna=False))
print(df["school"].unique())
print()
print(df["address"].value_counts(dropna=False))
print(df["address"].unique())
print()
print(df["famsize"].value_counts(dropna=False))
print(df["famsize"].unique())
print()
print(df["Pstatus"].value_counts(dropna=False))
print(df["Pstatus"].unique())

print(df["Fjob"].value_counts(dropna=False))
print(df["Fjob"].unique())

print(df["reason"].value_counts(dropna=False))
print(df["reason"].unique())
```

```python
binary_columns = [
    "sex",
    "school",
    "address",
    "famsize",
    "Pstatus",
    "schoolsup",
    "famsup",
    "activities",
    "nursery",
    "higher",
    "internet",
    "romantic",
    "math_paid",
    "por_paid"
]

onehot_columns = [
    "Mjob",
    "Fjob",
    "reason",
    "guardian"
]
```

```python
for col in binary_columns:
    encoder = OneHotEncoder(sparse_output=False,drop='if_binary')
    encoded = encoder.fit_transform(df[[col]])
    df[encoder.get_feature_names_out()] = encoded

for col in onehot_columns:
    encoder = OneHotEncoder(sparse_output=False)
    encoded = encoder.fit_transform(df[[col]])
    df[encoder.get_feature_names_out()] = encoded
```

```python
categorical_columns = binary_columns + onehot_columns
df.drop(columns=categorical_columns , inplace=True)
```
