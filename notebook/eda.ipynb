```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("data/student_performance.db")

# Mathematics
math_df = pd.read_csv("data/student-mat.csv")

math_df.to_sql(
    "math_students",
    connection,
    if_exists="replace",
    index=False
)

# Portuguese
por_df = pd.read_csv("data/student-por.csv")

por_df.to_sql(
    "portuguese_students",
    connection,
    if_exists="replace",
    index=False
)

connection.close()
```
