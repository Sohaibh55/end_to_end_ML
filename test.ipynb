```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'student_id': [101, 102, 103, 104],
    'sex': ['man', 'woman', 'woman', 'man'],
    'math_G1': [14, 18, 12, 15]
})
```

```python
# sparse_output=False returns a readable NumPy array instead of a sparse matrix
encoder = OneHotEncoder()

# Pass df[['sex']] as a 2D structure (DataFrame), not a 1D Series
df['sex'] = encoder.fit_transform(df[['sex']])

print("Categories learned:", encoder.categories_)
```
